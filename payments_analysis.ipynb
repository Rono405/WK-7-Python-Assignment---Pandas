# Payments Data Analysis

This notebook performs loading, analysis, and visualization of the Payments dataset following the Week 7 assignment requirements.

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")

# Load the dataset
try:
    df = pd.read_csv("Payments.csv")
    print("Data loaded successfully.")
except FileNotFoundError:
    print("Error: File not found.")
```

```
# Display the first few rows
df.head()
```

```
# Display dataset structure and missing values
df.info()
df.isnull().sum()
```

```
# Clean dataset by dropping missing values
df_clean = df.dropna()
df_clean.shape
```

## Basic Data Analysis

```
# Descriptive statistics
df_clean.describe()
```

```
# Grouping by 'Provider State' to compute mean payment amount
if 'Provider State' in df_clean.columns and 'Payment Amount' in df_clean.columns:
    group_mean = df_clean.groupby('Provider State')['Payment Amount'].mean().sort_values(ascending=False)
    group_mean
else:
    print("Required columns for grouping not found.")
```

## Visualizations

```
# Line chart: Payment over time
if 'Date of Payment' in df_clean.columns and 'Payment Amount' in df_clean.columns:
    df_clean['Date of Payment'] = pd.to_datetime(df_clean['Date of Payment'])
    time_series = df_clean.groupby('Date of Payment')['Payment Amount'].sum()
    time_series.plot(title='Total Payment Amount Over Time', figsize=(12,6))
    plt.xlabel('Date')
    plt.ylabel('Total Payment Amount')
    plt.tight_layout()
    plt.show()
```

```
# Bar chart: Top 10 states by average payment
if 'Provider State' in df_clean.columns:
    top_states = group_mean.head(10)
    top_states.plot(kind='bar', title='Top 10 States by Avg Payment Amount', figsize=(10,5))
    plt.xlabel('Provider State')
    plt.ylabel('Average Payment Amount')
    plt.tight_layout()
    plt.show()
```

```
# Histogram of Payment Amount
if 'Payment Amount' in df_clean.columns:
    plt.hist(df_clean['Payment Amount'], bins=30, edgecolor='black')
    plt.title('Distribution of Payment Amount')
    plt.xlabel('Payment Amount')
    plt.ylabel('Frequency')
    plt.tight_layout()
    plt.show()
```

```
# Scatter plot of two numerical columns
numeric_cols = df_clean.select_dtypes(include='number').columns
if len(numeric_cols) >= 2:
    plt.scatter(df_clean[numeric_cols[0]], df_clean[numeric_cols[1]], alpha=0.5)
    plt.title(f'Scatter Plot: {numeric_cols[0]} vs {numeric_cols[1]}')
    plt.xlabel(numeric_cols[0])
    plt.ylabel(numeric_cols[1])
    plt.tight_layout()
    plt.show()
else:
    print("Not enough numeric columns for scatter plot.")
```
